# 📘 Guide 01: Welcome to the guides

The guides are **self-study notebooks** that fill in background knowledge. You don't need to do them all, or in order. Dip in whenever a lab mentions something you'd like to understand better.

Think of them as the "refresher shelf" of the course: if you're new to the command line, Git or intermediate Python, a guide will get you comfortable in 15–30 minutes.

## 📚 The guides

| # | Guide | Read it if… |
|---|---|---|
| 01 | Welcome (this one) | you're starting the course |
| 02 | The command line | terminals feel unfamiliar |
| 03 | Git and GitHub | you've never cloned, pulled or opened a Pull Request |
| 04 | Technical foundations | environment variables, APIs, HTTP and JSON are fuzzy |
| 05 | Notebooks | you're new to Jupyter |
| 06 | Python foundations | you'd like a refresher on functions, classes, comprehensions |
| 07 | AI-assisted coding and debugging | you want to get the most from Cursor / Claude Code |
| 08 | Debugging | you'd like a method for when things go wrong |
| 09 | AI APIs, costs and Ollama | you want to understand pricing or run models for free |
| 10 | Intermediate Python | type hints and **Pydantic** (used heavily from week 1, day 3) |
| 11 | Async Python | `async` / `await` (used in weeks 3–6) |
| 12 | Starting your own project | you want to build your own variant of GradeWise |
| 13 | Streamlit and FastAPI | you'd like a head start on week 6 |
| 14 | Docker basics | containers are new to you |
