# 📘 Guide 02: The command line

**Who it's for:** anyone who feels a little nervous when a lab says "open a terminal and run…". **Time:** about 20 minutes.

The command line is just a text conversation with your computer. You type a command, press Enter, it answers. Once you know a dozen commands, you'll find it faster than clicking around, and every tool in this course (`uv`, `git`, `streamlit`, `docker`) lives there.

### 🎯 You'll learn
1. How to open a terminal on Windows, Mac and Linux (including inside Cursor / VS Code)

3. Creating, copying and deleting files and folders
4. Environment variables and the `PATH`
5. Running the course tools with `uv run`

## Part 2: Where am I, and what's here?

| Task | Windows (PowerShell) | Mac / Linux |
|---|---|---|
| Print the current folder | `pwd` | `pwd` |
| List files | `ls` (or `dir`) | `ls` |
| List including hidden files | `ls -Force` | `ls -a` |
| Go into a folder | `cd week1` | `cd week1` |
| Go up one level | `cd ..` | `cd ..` |
| Go to your home folder | `cd ~` | `cd ~` |
| Clear the screen | `cls` (or `clear`) | `clear` |

Good news: PowerShell understands the most common Unix-style names (`ls`, `pwd`, `cd`, `cat`), so most of the time the same command works everywhere.

Two keyboard tricks that will save you hours:
- **Tab** auto-completes a file or folder name. Type `cd we` then Tab → `cd week1`.
- **Up arrow** brings back your previous command.

A notebook can tell you the same things using Python, which works identically on every OS:

````python
from pathlib import Path

here = Path.cwd()
print(f"📍 This notebook is running in: {here}")
print("\n📂 What's next to it:")
for item in sorted(here.parent.iterdir())[:15]:
    kind = "📁" if item.is_dir() else "📄"
    print(f"   {kind} {item.name}")
````

## Part 3: Paths: absolute, relative, and the slash question

- An **absolute path** starts from the top of the drive: `C:\Users\emma\agentic-rag-gradewise\week1` on Windows, `/Users/emma/agentic-rag-gradewise/week1` on a Mac.
- A **relative path** starts from where you are: `week1/day1.ipynb`, or `../data` ("up one level, then into data").
- Windows uses `\` and Mac/Linux use `/`. PowerShell happily accepts `/` too, so the course docs use `/` everywhere.

In Python code, we never glue paths together with strings. We use `pathlib`, which picks the right slash for you:

````python
from pathlib import Path

repo = Path.cwd().parent
exam_folder = repo / "data" / "exams"  # 💡 the / operator joins paths, on every OS

print("Joined path:   ", exam_folder)
print("Relative to me:", exam_folder.relative_to(repo))
print("Exists?        ", exam_folder.exists())
print("Parts:         ", exam_folder.parts[-2:])
````

## Part 4: Making, copying, moving and deleting

| Task | Windows (PowerShell) | Mac / Linux |
|---|---|---|
| Make a folder | `mkdir scratch` | `mkdir scratch` |
| Make nested folders | `mkdir scratch/a/b` (creates parents automatically) | `mkdir -p scratch/a/b` |
| Create an empty file | `New-Item notes.txt` | `touch notes.txt` |
| Show a file's contents | `cat notes.txt` | `cat notes.txt` |
| Copy | `cp notes.txt backup.txt` | `cp notes.txt backup.txt` |
| Move / rename | `mv notes.txt old.txt` | `mv notes.txt old.txt` |
| Delete a file | `rm old.txt` | `rm old.txt` |
| Delete a folder and everything in it | `rm -r -Force scratch` | `rm -rf scratch` |

## Part 5: Environment variables and the PATH

An **environment variable** is a named value your shell passes to every program it starts. Two matter to us:
- `PATH`: the list of folders the shell searches when you type a command like `uv` or `git`. "Command not found" almost always means "it's not on your PATH".
- Your API keys. The course keeps them in a `.env` file instead (see Guide 04), but it's the same idea.

| Task | Windows (PowerShell) | Mac / Linux |
|---|---|---|
| Show one variable | `$env:PATH` | `echo $PATH` |
| Set one for this session | `$env:DATA_SIZE = "small"` | `export DATA_SIZE=small` |
| Where does a command live? | `Get-Command uv` | `which uv` |

Python sees the same variables:

````python
import os

path_entries = os.environ.get("PATH", "").split(os.pathsep)  # 💡 os.pathsep is ';' on Windows, ':' elsewhere
print(f"Your PATH has {len(path_entries)} folders. The first three:")
for p in path_entries[:3]:
    print("  ", p)

print("\nHome folder:", os.path.expanduser("~"))
````
