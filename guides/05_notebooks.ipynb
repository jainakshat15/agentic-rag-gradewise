## Part 1: Cells and the kernel

- A **markdown cell** holds text like this one. Double-click to edit, `Shift + Enter` to render.
- A **code cell** holds Python. `Shift + Enter` runs it and moves to the next cell.
- The **kernel** is a Python process running in the background. It remembers every variable you've created, **in the order you ran cells**, not the order they appear.

The number in `[ ]` next to a cell shows the order it ran in. Run these two cells:

```python
marks = [4, 3, 5]
print("Created marks:", marks)
```

```python
marks.append(2)  # 🧪 TRY: run this cell three times, then look at the output
print("Marks now:", marks, "| total:", sum(marks))
```

## Part 2: The hidden state trap

Run the second cell above a few times and the list keeps growing. The notebook *looks* the same, but the kernel's memory has changed. This is the #1 cause of "it worked a minute ago!"

The cure is simple: **Restart** the kernel and **Run All**. In Cursor / VS Code, both buttons are in the notebook toolbar. If a notebook works after Restart + Run All, it really works.

## Part 3: Shortcuts and magics

Press `Esc` to enter *command mode* (the cell border changes), then:

| Key | Action |
|---|---|
| `A` / `B` | insert cell above / below |
| `M` / `Y` | turn cell into markdown / code |
| `D D` | delete cell |
| `Shift + Enter` | run and move on |

(On a Mac, `Cmd` replaces `Ctrl` in editor shortcuts.)

**Magics** are notebook-only helpers that start with `%`. `!` sends a line to the shell:

```python
%time total = sum(range(1_000_000))
print(total)
```

## Part 4: Rich output

The last expression in a cell is displayed automatically. For nicer output we use `IPython.display`, which is how labs render LLM answers as formatted Markdown:

```python
from IPython.display import Markdown, display

from gradewise.evals.compat import import_pandas

pd = import_pandas()  # 💡 same as `import pandas as pd`, plus a fix for locked-down Windows machines

display(Markdown("### 📝 Feedback for Sam\n**Q2:** Good definition of a stack. Mention *LIFO* explicitly for full marks."))

pd.DataFrame(
    {"student": ["Sam", "Priya", "Leo"], "Q1": [4, 5, 3], "Q2": [3.5, 4, 2]},
)
```

## Part 5: Notebooks vs `.py` files

Notebooks are for **exploring and explaining**. Once code is reusable, we move it into the `gradewise/` package and import it. That's why labs say `from gradewise.models import chat_model` instead of repeating setup everywhere.

A good rule: if you've copy-pasted a function into a second notebook, it belongs in a `.py` file.

And before sharing a notebook: **Clear All Outputs** (see Guide 03 for `nbstripout`).
