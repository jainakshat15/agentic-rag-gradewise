# 📘 Guide 03: Git and GitHub

**Who it's for:** you've never cloned a repo, or Git still feels like magic spells. **Time:** about 30 minutes.

**Git** keeps a full history of a folder, so you can always go back. **GitHub** is a website that hosts Git repos so people can share them. You need just a handful of commands to follow the course, get my updates, and share your own GradeWise variant with other students.

### 🎯 You'll learn
1. The mental model: working folder → staging → commits → remote
2. `clone` and `pull`: getting the course and its updates
3. Branches: experimenting safely
4. Fork + Pull Request: sharing your work in `community_contributions/`
5. Keeping notebooks tidy with `nbstripout`

## Part 1: The mental model

```
 your edits          git add          git commit          git push
 working folder  ───────────▶  staging  ───────────▶  local history  ───────────▶  GitHub (remote)
                                                                   ◀───────────
                                                                     git pull
```

- A **commit** is a saved snapshot with a message, like "Add rubric for Q3".
- **Staging** (`git add`) is choosing which changes go into the next snapshot.
- The **remote** (usually called `origin`) is the copy on GitHub.

Commands are identical on Windows, Mac and Linux. Let's check Git is installed:

````python
!git --version
````

The first time you use Git on a machine, tell it who you are (this goes into every commit):

```bash
git config --global user.name "Emma Brooks"
git config --global user.email "emma@example.com"
```

## Part 2: Getting the course: clone and pull

```bash
cd ~/projects        # Windows: cd ~\projects   (any folder you like)
git clone https://github.com/<course-owner>/agentic-rag-gradewise.git
cd agentic-rag-gradewise
```

I'll keep improving the repo while the course runs. To get the latest:

```bash
git pull
uv sync              # in case new packages were added
```

If `git pull` complains that your local changes would be overwritten, that's usually because you edited a lab notebook. Two easy fixes:
- Work on **copies**: duplicate `day3.ipynb` as `day3_mine.ipynb` and edit that. Pulls never touch it.
- Or put your changes aside, pull, and bring them back: `git stash`, `git pull`, `git stash pop`.

Here are the read-only commands for looking around. They're safe to run any time:

````python
!git status --short --branch
!git log --oneline -5
````

## Part 3: Your everyday loop

```bash
git status                          # what changed?
git diff                            # show me the exact lines
git add week1/day3_mine.ipynb       # stage one file (or: git add .  for everything)
git commit -m "Try a stricter rubric prompt"
git push                            # send commits to GitHub (your fork, see Part 5)
```

## Part 4: Branches: a safe place to experiment

A branch is a parallel line of history. Make one before trying something risky; if it doesn't work, switch back and nothing is lost.

```bash
git switch -c moderation-idea        # create a branch and move onto it
# ... edit, add, commit ...
git switch main                      # back to the original
git branch                           # list branches (* marks the current one)
git merge moderation-idea            # bring the experiment into main when you're happy
```

Python can ask Git questions too. This is handy in scripts:

````python
import subprocess

result = subprocess.run(["git", "branch", "--show-current"], capture_output=True, text=True)
if result.returncode == 0:
    print(f"🌿 You are on branch: {result.stdout.strip() or '(detached)'}")
else:
    print("ℹ️ Not inside a git repository:", result.stderr.strip())
````

## Part 5: Sharing your work: fork + Pull Request

You can't push to my repo directly, so GitHub uses **forks**: your own copy of the repo under your account.

1. On the course's GitHub page, click **Fork**. You now have `github.com/<you>/agentic-rag-gradewise`.
2. Clone **your fork** (or point your existing clone at it):
   ```bash
   git remote rename origin upstream
   git remote add origin https://github.com/<you>/agentic-rag-gradewise.git
   ```
3. Make a branch and put your work **only** in `community_contributions/<your-github-username>/`, with a short `README.md`.
4. Commit and push: `git push -u origin my-history-grader`.
5. On GitHub, click **Compare & pull request**. Use a title like `Week 5: History essay grader by @sam`.

To get my newest changes into your fork later: `git pull upstream main`.

The full checklist is in `CONTRIBUTING.md` at the repo root.

## Part 6: Clean notebooks with nbstripout

Notebook outputs (printed text, images, your API key prefix!) are saved inside the `.ipynb` file. They make diffs huge and pull requests hard to review. **nbstripout** removes outputs automatically whenever you commit:

```bash
uv run nbstripout --install     # once per clone; afterwards every commit is stripped automatically
```

Prefer to do it by hand? In Cursor / VS Code: the `…` menu at the top of the notebook → **Clear All Outputs**, then save.
