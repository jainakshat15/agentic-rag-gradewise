## Part 1: f-strings, lists and dicts

```python
student = "Sam"
marks = {"Q1": 4, "Q2": 3.5, "Q3": 2}  # dict: question → marks
max_marks = {"Q1": 5, "Q2": 5, "Q3": 5}

total = sum(marks.values())
out_of = sum(max_marks.values())
print(f"{student} scored {total}/{out_of} = {total / out_of:.0%}")  # 💡 :.0% formats as a percentage

for question, got in marks.items():
    print(f"  {question}: {got:>4} / {max_marks[question]}")
```

## Part 2: Functions

Default values and keyword arguments make functions flexible. Labs use them a lot, for example `chat_model(temperature=0)`.

```python
def percentage(got: float, out_of: float, *, decimals: int = 1) -> float:
    """Return got/out_of as a percentage. The * forces decimals to be passed by name."""
    if out_of <= 0:
        raise ValueError("out_of must be positive")
    return round(100 * got / out_of, decimals)


print(percentage(9.5, 15))
print(percentage(9.5, 15, decimals=3))
```

## Part 3: Comprehensions

A comprehension builds a list or dict in one readable line. Read it as "give me *this* for each item in *that*, if *condition*".

```python
scripts = [
    {"student": "Sam", "total": 9.5},
    {"student": "Priya", "total": 14},
    {"student": "Leo", "total": 6},
]

names = [s["student"] for s in scripts]
needs_review = [s["student"] for s in scripts if s["total"] < 8]
by_name = {s["student"]: s["total"] for s in scripts}

print(names)
print("Asha should review:", needs_review)
print(by_name)
```

## Part 4: Classes and dataclasses

A class bundles data with behaviour. A `dataclass` writes the boring parts (`__init__`, `__repr__`) for you. In the course we mostly use **Pydantic** models, which are dataclasses with validation (Guide 10).

```python
from dataclasses import dataclass, field


@dataclass
class Script:
    student: str
    marks: dict[str, float] = field(default_factory=dict)  # ⚠️ never use a bare {} as a default

    @property
    def total(self) -> float:
        return sum(self.marks.values())

    def add_mark(self, question: str, value: float) -> None:
        self.marks[question] = value


sam = Script("Sam")
sam.add_mark("Q1", 4)
sam.add_mark("Q2", 3.5)
print(sam, "→ total", sam.total)
```

## Part 5: Files and errors

`pathlib` reads and writes text in one line. Always pass `encoding="utf-8"`: Windows otherwise picks a different default, and accented student names get mangled.

```python
import tempfile
from pathlib import Path

folder = Path(tempfile.mkdtemp())
answer_file = folder / "sam_q2.txt"
answer_file.write_text("A stack is last-in, first-out. Ça marche!", encoding="utf-8")
print("Read back:", answer_file.read_text(encoding="utf-8"))

try:
    (folder / "missing.txt").read_text(encoding="utf-8")
except FileNotFoundError as e:
    print(f"😅 Handled it: {type(e).__name__}: {e.filename}")
```
