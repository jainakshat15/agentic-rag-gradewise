# 📘 Guide 08: Debugging

**Who it's for:** anyone who's ever stared at a red error and felt stuck. That's all of us. **Time:** about 25 minutes.

Debugging is a skill, not a talent. With a calm, repeatable method you'll fix most problems in minutes.

### 🎯 You'll learn
1. How to read a traceback
2. A five-step debugging method
3. Inspecting values with `print`, `repr` and `breakpoint`
4. The usual suspects in this course (and their fixes)

## Part 1: Reading a traceback

A traceback lists the calls that led to the error, **oldest first**. The last line says *what* went wrong; the line just above shows *where*. Read it from the bottom up.

````python
import traceback


def total_marks(grades: list[dict]) -> float:
    return sum(g["marks"] for g in grades)


def report(student: str, grades: list[dict]) -> str:
    return f"{student}: {total_marks(grades)}"


grades = [{"question": "Q1", "marks": 4}, {"question": "Q2", "mark": 3}]  # ⚠️ spot the typo

try:
    report("Sam", grades)
except KeyError:
    traceback.print_exc()  # 💡 prints the traceback without stopping the notebook
````

Reading bottom-up: `KeyError: 'marks'` → raised in `total_marks`, inside the generator `g["marks"]` → called from `report`. So one of the dicts lacks a `"marks"` key. The second one says `"mark"`.

## Part 2: The method

1. **Read** the whole error, slowly. Most errors tell you exactly what's wrong.
2. **Reproduce** it reliably, in the smallest piece of code you can.
3. **Inspect**: print the actual values just before the failing line. Your assumption is usually what's wrong.
4. **Hypothesise and change one thing** at a time.
5. **Confirm** the fix, and ideally add a test so it never returns.

## Part 3: Inspecting values

`repr()` (or `!r` in an f-string) shows a value *exactly*, including hidden whitespace and types. It catches a surprising number of bugs:

````python
score_from_llm = " 4\n"  # what a model might return as text
print(f"print shows:  {score_from_llm}")
print(f"repr shows:   {score_from_llm!r}")
print(f"type:         {type(score_from_llm).__name__}")
print(f"fixed:        {float(score_from_llm.strip())!r}")

for g in grades:
    print(sorted(g.keys()))  # 💡 quickly compare the shape of each item
````

For a closer look, put `breakpoint()` on a line in a script and run it: Python pauses there and you can type variable names to inspect them (`n` next line, `c` continue, `q` quit). In Cursor / VS Code, you can also click left of a line number to set a red-dot breakpoint and use **Debug Cell** on a notebook cell.

```python
def total_marks(grades):
    breakpoint()          # execution stops here
    return sum(g["marks"] for g in grades)
```

## Part 4: The usual suspects

| Symptom | Likely cause | Fix |
|---|---|---|
| `ModuleNotFoundError: No module named 'langchain'` | wrong kernel / not using `.venv` | pick the `.venv` kernel; run `uv sync` |
| `AuthenticationError` / `401` | key missing, wrong, or `.env` not saved | check `.env`, re-run `load_dotenv(override=True)` |
| `RateLimitError` / `429` | too many requests, or no credit | wait and retry; check billing; try `BUDGET_MODE=true` |
| `UnicodeDecodeError` (Windows) | file read without UTF-8 | `read_text(encoding="utf-8")` |
| `'uv' is not recognized` / `command not found` | uv not on PATH | restart the terminal; see Guide 02 |
| Notebook works, then doesn't | hidden state | Restart + Run All (Guide 05) |
| `ValidationError` from Pydantic | data doesn't match the schema | read which field; print the raw data (Guide 10) |

GradeWise's own errors (`gradewise/errors.py`) always say **what** happened, the likely **cause**, and a **fix**. Read them fully!
