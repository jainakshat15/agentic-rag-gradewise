# 📘 Guide 12: Starting your own project

**Who it's for:** you've got an idea (grading history essays, marking maths homework, checking lab reports) and want to build your own variant of GradeWise. **Time:** about 25 minutes.

The best way to lock in what you learn is to build something of your own. This guide gets you from blank folder to a clean, working skeleton.

### 🎯 You'll learn
1. Scoping an idea so you can actually finish it
2. Creating a new project with `uv`
3. A sensible folder layout
4. Settings, secrets and a first test
5. Sharing it with the community

## Part 1: Scope it small

Write your idea as one sentence: **"For *who*, take *input*, produce *output*, judged against *source of truth*."**

> *For Mr. Miller's Year 10 history class, take typed essays, produce marks and feedback per criterion, judged against the exam board's mark scheme.*

Then pick the smallest version that's useful: one question type, five sample answers, one rubric. You can always grow it.

Data: use **synthetic** or properly anonymised answers. Never real student data without permission.

## Part 2: Create the project with uv

The commands are the same on every OS (on Windows, run them in PowerShell):

```bash
cd ~/projects                     # Windows: cd ~\projects
uv init essaywise --python 3.12   # creates pyproject.toml, README.md, a sample main.py
cd essaywise
uv add langchain langchain-openai pydantic-settings python-dotenv
uv add --dev pytest ruff ipykernel
uv run python main.py
```

Then open the folder in Cursor and pick `.venv` as your notebook kernel.

## Part 3: A layout that grows well

Here's a skeleton modelled on this repo. The cell below builds it in a temporary folder so you can see it (nothing is written into the course repo):

````python
import tempfile
from pathlib import Path

root = Path(tempfile.mkdtemp()) / "essaywise"
layout = {
    "pyproject.toml": "[project]\nname = 'essaywise'\nrequires-python = '>=3.12'\n",
    ".env.example": "OPENROUTER_API_KEY=\nCHAT_MODEL=openrouter:openai/gpt-6-luna\n",
    ".gitignore": ".env\n.venv/\noutput/\n__pycache__/\n",
    "README.md": "# EssayWise\n",
    "essaywise/__init__.py": "",
    "essaywise/config.py": "# settings (pydantic-settings)\n",
    "essaywise/schemas.py": "# Pydantic models: Essay, Criterion, Grade\n",
    "essaywise/grader.py": "# the core logic\n",
    "data/samples/essay_01.txt": "The causes of the war were...\n",
    "notebooks/01_explore.ipynb": "{}",
    "tests/test_schemas.py": "def test_placeholder():\n    assert True\n",
}
for rel, content in layout.items():
    path = root / rel
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(content, encoding="utf-8")

for path in sorted(root.rglob("*")):
    depth = len(path.relative_to(root).parts) - 1
    print("    " * depth + ("📁 " if path.is_dir() else "📄 ") + path.name)
````

Key ideas:
- **`.env.example`** is committed (empty values) so others know which settings exist; **`.env`** is not.
- Reusable logic goes in the package (`essaywise/`); notebooks import from it.
- `output/` is git-ignored: generated files never clutter history.

## Part 4: Settings and a first test

Copy the pattern from `gradewise/config.py`: one `Settings` class, model names only in `.env`. And write a test on day one: it's much easier to keep a project healthy than to rescue it.

````python
from pydantic import BaseModel, Field


class Criterion(BaseModel):
    name: str
    max_marks: int = Field(gt=0)


def total_available(criteria: list[Criterion]) -> int:
    return sum(c.max_marks for c in criteria)


# 🧪 this is what tests/test_schemas.py might check
def test_total_available():
    assert total_available([Criterion(name="Argument", max_marks=6), Criterion(name="Evidence", max_marks=4)]) == 10


test_total_available()
print("✅ test passed")
````

Run tests in your project with `uv run python -m pytest -q`, and lint with `uv run python -m ruff check .`.
